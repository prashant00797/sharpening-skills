# 🧩 Q1. User Profile Endpoint with Conditional Response

## 📌 Scenario

You are building a **user profile API** for a healthcare app.

Create a `GET` endpoint:

```
/users/{user_id}
```

### Requirements:

* Accept:

  * `user_id` as a **path parameter**
  * `include_health_data` as an **optional query parameter** (default: `False`)
* Use **mock data** (Python dictionary), no database

---

## 🎯 Expected Behavior

* If `include_health_data = False`
  → Return **basic user info**

* If `include_health_data = True`
  → Return **full user info including health data**

---

## 🧱 Response Models

### 🟢 UserBasic

* `name`
* `email`

---

### 🔴 UserFull

* `name`
* `email`
* `blood_group`
* `allergies`

---

## 🧠 Concepts Tested

* Path Parameters
* Query Parameters
* Pydantic Response Models
* Conditional Response Handling

---

## 💡 Hint

* Define **two separate Pydantic models**
* Use an `if/else` inside the endpoint:

  * return basic model OR full model
* Path param → part of URL
* Query param → function argument with default value

---

### 🧩 Q2. Create Product Endpoint 

#### 📌 What you’re building

A **POST API endpoint** that creates a new product in an e-commerce system.

---

### 🧱 1. Request Body Structure (Input)

You need a **Pydantic model** to define and validate incoming data.

**Fields:**

* `name` → string → **required**
* `price` → float → **required**, must be **> 0**
* `category` → string → **required**
* `description` → string → optional

👉 Purpose:

* Ensures correct data types
* Automatically validates constraints (like price > 0)

---

### ⚙️ 2. Endpoint Behavior

When a request hits `POST /products`:

* Accept JSON body matching your model
* Validate input (handled by FastAPI + Pydantic)
* Generate additional fields (not provided by user)

---

### 🆔 3. Auto-generated Fields

You must add:

* `id`

  * Use **UUID4**
  * Ensures uniqueness
* `created_at`

  * Use **UTC timestamp**
  * Convert to **ISO format string**

---

### 📤 4. Response

Return a **complete product object** including:

* All input fields (`name`, `price`, etc.)
* Generated fields (`id`, `created_at`)

---

### ✅ 5. HTTP Status Code

* Use: **201 Created**
* This tells the client:

  > “A new resource has been successfully created”

---

### 🧠 What this question is actually testing

* How request bodies work in FastAPI
* Data validation using Pydantic
* Proper REST semantics (201 vs 200)
* Generating backend-controlled fields
* Understanding of UUID and timestamps
